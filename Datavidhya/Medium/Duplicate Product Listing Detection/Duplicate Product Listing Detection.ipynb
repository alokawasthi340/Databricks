# Duplicate Product Listing Detection

## Problem Statement

You are given a table named `duplicate_product_listing` that contains product listings from various vendors on an e-commerce platform.

Your task is to identify vendors who have posted duplicate product listings.

A duplicate listing is defined as two or more listings from the same vendor that have the same product name and description.

## Input Table

### duplicate_product_listing

| Column Name | Data Type |
|------------|-----------|
| listing_id | INT |
| vendor_id | INT |
| product_name | VARCHAR |
| description | VARCHAR |

## Requirements

- Analyze product listings across vendors.
- Determine which vendors have duplicate product listings.
- A duplicate listing is based on matching product name and description for the same vendor.
- Handle NULL values appropriately.
- Return results matching the required output schema.

## Output Columns

| Column Name |
|------------|
| duplicate_vendors |

## Sample Input

### duplicate_product_listing

| listing_id | vendor_id | product_name | description |
|------------|------------|------------|------------|
| 101 | 501 | Bluetooth Speaker | A portable speaker with Blueto |
| 102 | 501 | Bluetooth Speaker | A portable speaker with Blueto |
| 103 | 502 | Wireless Earbuds | Comfortable earbuds with high- |
| 104 | 502 | Wireless Earbuds | Comfortable earbuds with high- |

## Expected Output Schema

| Column Name | Data Type |
|------------|-----------|
| duplicate_vendors | INT |

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType
from pyspark.sql.functions import *

duplicate_product_listing_schema = StructType([
    StructField("listing_id", IntegerType(), True),
    StructField("vendor_id", IntegerType(), True),
    StructField("product_name", StringType(), True),
    StructField("description", StringType(), True)
])

duplicate_product_listing_data = [
    (101, 501, "Bluetooth Speaker", "A portable speaker with Blueto"),
    (102, 501, "Bluetooth Speaker", "A portable speaker with Blueto"),
    (103, 502, "Wireless Earbuds", "Comfortable earbuds with high-"),
    (104, 502, "Wireless Earbuds", "Comfortable earbuds with high-")
]

duplicate_product_listing_df = spark.createDataFrame(
    duplicate_product_listing_data,
    schema=duplicate_product_listing_schema
)
```

```python
result_df = (
    duplicate_product_listing_df.groupBy("vendor_id", "product_name", "description")
    .agg(count("*").alias("count"))
    .filter(col("count") > 1)
    .select(count("*").alias("duplicate_vendors"))
)

display(result_df)
```
