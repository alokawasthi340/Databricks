# Top Spotify Artist Rankings

## Problem Statement

Spotify wants to rank artists based on the number of appearances their songs make in the global Top 10 rankings.

Determine artist rankings using song appearance information and return artists whose ranking falls within the top five positions.

## Input Tables

### ar_artists

| Column Name | Data Type |
|------------|-----------|
| artist_id | INT |
| artist_name | VARCHAR |
| label_owner | VARCHAR |

### ar_songs

| Column Name | Data Type |
|------------|-----------|
| song_id | INT |
| artist_id | INT |
| name | VARCHAR |

### ar_global_song_rank

| Column Name | Data Type |
|------------|-----------|
| day | INT |
| song_id | INT |
| rank | INT |

## Requirements

- Consider song ranking information from the global rankings table.
- Top 10 rankings are inclusive.
- Count qualifying song appearances for each artist.
- Rank artists based on appearance counts.
- Use artist name as the tie-breaker when required.
- Return artists whose rank is at most 5.
- Return results ordered by `artist_rank` ascending.
- Return results matching the required output schema and order.

## Output Columns

| Column Name |
|------------|
| artist_name |
| artist_rank |

## Sample Input

### ar_artists

| artist_id | artist_name | label_owner |
|------------|------------|-------------|
| 101 | Ed Sheeran | Warner Music Group |
| 120 | Drake | Warner Music Group |

### ar_songs

| song_id | artist_id | name |
|----------|----------|------|
| 45202 | 101 | Shape of You |
| 19960 | 120 | Hotline Bling |

### ar_global_song_rank

| day | song_id | rank |
|-----|---------|------|
| 1 | 45202 | 5 |
| 3 | 45202 | 2 |
| 1 | 19960 | 3 |
| 9 | 19960 | 15 |
| 2 | 30001 | 10 |

## Sample Output

| artist_name | artist_rank |
|------------|------------|
| Ed Sheeran | 1 |
| Bad Bunny | 2 |
| Drake | 3 |

## Expected Output Schema

| Column Name | Data Type |
|------------|-----------|
| artist_name | STRING |
| artist_rank | INT |

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *

# ar_artists
ar_artists_schema = StructType([
    StructField("artist_id", IntegerType(), True),
    StructField("artist_name", StringType(), True),
    StructField("label_owner", StringType(), True)
])

ar_artists_data = [
    (101, "Ed Sheeran", "Warner Music Group"),
    (120, "Drake", "Warner Music Group")
]

ar_artists_df = spark.createDataFrame(
    ar_artists_data,
    schema=ar_artists_schema
)

# ar_songs
ar_songs_schema = StructType([
    StructField("song_id", IntegerType(), True),
    StructField("artist_id", IntegerType(), True),
    StructField("name", StringType(), True)
])

ar_songs_data = [
    (45202, 101, "Shape of You"),
    (19960, 120, "Hotline Bling")
]

ar_songs_df = spark.createDataFrame(
    ar_songs_data,
    schema=ar_songs_schema
)

# ar_global_song_rank
ar_global_song_rank_schema = StructType([
    StructField("day", IntegerType(), True),
    StructField("song_id", IntegerType(), True),
    StructField("rank", IntegerType(), True)
])

ar_global_song_rank_data = [
    (1, 45202, 5),
    (3, 45202, 2),
    (1, 19960, 3),
    (9, 19960, 15),
    (2, 30001, 10)
]

ar_global_song_rank_df = spark.createDataFrame(
    ar_global_song_rank_data,
    schema=ar_global_song_rank_schema
)
```
