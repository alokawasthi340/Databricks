## Problem Statement
You’re working as a Data Engineer at a company that builds Customer Relationship Management (CRM) software. Your goal is to build a unified view that shows order details along with customer and product information for internal dashboards and reporting.

You are given three datasets (or tables):
- `crm_customers`
- `crm_orders`
- `crm_products`

### Required Output Columns
- order_id
- customer_name
- customer_email
- product_name
- product_category
- order_date

### Sorting Requirement
Sort the final result by `order_id` in ascending order.

> Note: Solution logic intentionally not included.


````python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

# -------------------------
# crm_customers DataFrame
# -------------------------
customers_data = [
    (1, "John", "Doe", "john.doe@email.com"),
    (2, "Jane", "Smith", "jane.smith@email.com")
]

customers_df = spark.createDataFrame(
    customers_data,
    ["customer_id", "first_name", "last_name", "email"]
)

customers_df.show()


# -------------------------
# crm_orders DataFrame
# -------------------------
orders_data = [
    (1001, 1, 101, "2023-01-10"),
    (1002, 2, 102, "2023-01-11")
]

orders_df = spark.createDataFrame(
    orders_data,
    ["order_id", "customer_id", "product_id", "order_date"]
)

orders_df.show()


# -------------------------
# crm_products DataFrame
# -------------------------
products_data = [
    (101, "Product A", "Category1"),
    (102, "Product B", "Category2")
]

products_df = spark.createDataFrame(
    products_data,
    ["product_id", "product_name", "category"]
)

products_df.show()
````

````python
from pyspark.sql.functions import *

result_df = (
    orders_df.join(customers_df, orders_df.customer_id == customers_df.customer_id)
    .join(products_df, orders_df.product_id == products_df.product_id)
    .select(
        orders_df.order_id,
        concat(customers_df.first_name, lit(" "), customers_df.last_name).alias(
            "customer_name"
        ),
        customers_df.email.alias("customer_email"),
        products_df.product_name,
        products_df.category.alias("product_category"),
        orders_df.order_date,
    )
)
display(result_df)
````
