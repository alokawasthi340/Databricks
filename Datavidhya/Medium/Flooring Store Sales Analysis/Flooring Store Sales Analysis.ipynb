# Flooring Store Sales Analysis

## Problem Statement

You are a data analyst at the multinational flooring retailer Floors'R'Us.

The sales system stores customer names in a single `full_name` field and product attributes in a comma-separated `product_info` field. The sales team requires a clean order report with customer and product details organized into separate fields.

Build one row per order using the customer, order, and product datasets.

## Input Tables

### fr_customers

| Column Name | Data Type |
|------------|-----------|
| customer_id | INT |
| full_name | VARCHAR |
| location | VARCHAR |

### fr_orders

| Column Name | Data Type |
|------------|-----------|
| order_id | INT |
| customer_id | INT |
| product_id | INT |
| quantity | INT |

### fr_products

| Column Name | Data Type |
|------------|-----------|
| product_id | INT |
| product_info | VARCHAR |

## Requirements

- Build one row per order.
- Combine order, customer, and product information.
- Derive `first_name` and `last_name` from `full_name`.
- Derive `product_type` and `product_color` from `product_info`.
- Only orders with matching customer and product records should be included.
- Return rows in ascending order of `order_id`.

## Output Columns

Return columns in exactly the following order:

1. customer_id
2. first_name
3. last_name
4. location
5. order_id
6. product_type
7. product_color
8. product_id
9. quantity

## Sample Input

### fr_customers

| customer_id | full_name | location |
|------------|-----------|----------|
| 1 | John Doe | Texas |
| 2 | Jane Smith | California |

### fr_orders

| order_id | customer_id | product_id | quantity |
|----------|-------------|------------|----------|
| 1001 | 1 | 101 | 5 |
| 1002 | 2 | 102 | 2 |

### fr_products

| product_id | product_info |
|------------|--------------|
| 101 | Carpet,Red |
| 102 | Tile,Blue |

## Expected Output Schema

| Column Name | Data Type |
|------------|-----------|
| customer_id | INT |
| first_name | STRING |
| last_name | STRING |
| location | STRING |
| order_id | INT |
| product_type | STRING |
| product_color | STRING |
| product_id | INT |
| quantity | INT |

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType
from pyspark.sql.functions import *
# fr_customers
fr_customers_schema = StructType([
    StructField("customer_id", IntegerType(), True),
    StructField("full_name", StringType(), True),
    StructField("location", StringType(), True)
])

fr_customers_data = [
    (1, "John Doe", "Texas"),
    (2, "Jane Smith", "California")
]

fr_customers_df = spark.createDataFrame(
    fr_customers_data,
    schema=fr_customers_schema
)

# fr_orders
fr_orders_schema = StructType([
    StructField("order_id", IntegerType(), True),
    StructField("customer_id", IntegerType(), True),
    StructField("product_id", IntegerType(), True),
    StructField("quantity", IntegerType(), True)
])

fr_orders_data = [
    (1001, 1, 101, 5),
    (1002, 2, 102, 2)
]

fr_orders_df = spark.createDataFrame(
    fr_orders_data,
    schema=fr_orders_schema
)

# fr_products
fr_products_schema = StructType([
    StructField("product_id", IntegerType(), True),
    StructField("product_info", StringType(), True)
])

fr_products_data = [
    (101, "Carpet,Red"),
    (102, "Tile,Blue")
]

fr_products_df = spark.createDataFrame(
    fr_products_data,
    schema=fr_products_schema
)
```

```python
result_df = (
    fr_customers_df.join(fr_orders_df, on="customer_id", how="inner")
    .join(fr_products_df, on="product_id", how="inner")
    .select(
        col("customer_id"),
        split(col("full_name"), " ")[0].alias("first_name"),
        split(col("full_name"), " ")[1].alias("last_name"),
        col("location"),
        col("order_id"),
        split(col("product_info"), ",")[0].alias("product_type"),
        split(col("product_info"), ",")[1].alias("product_color"),
        col("product_id"),
        col("quantity"),
    )
)
display(result_df)
```
