```sql
CREATE OR REPLACE TABLE ccd_credit_cards (
    issue_month INT,
    issue_year INT,
    card_name STRING,
    issued_amount INT
);

INSERT INTO ccd_credit_cards VALUES
(1, 2021, 'Sapphire Plus', 170000),
(2, 2021, 'Sapphire Plus', 175000),
(3, 2021, 'Sapphire Plus', 180000),
(3, 2021, 'Freedom Flex', 65000);

select
card_name,
issued_amount
from
(
select 
card_name,
issued_amount,
dense_rank() over (partition by card_name order by issue_year ,issue_month) as rnk
from ccd_credit_cards
) x
where x.rnk=1
order by issued_amount desc;
```
