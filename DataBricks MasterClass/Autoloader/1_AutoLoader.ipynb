```python
%sql
create volume if not exists databrickscatalog.bronze.autovol
```

```python
df=spark.readStream.format("cloudFiles") \
.option("cloudFiles.format", "csv") \
.option("cloudFiles.schemaLocation","/Volumes/databrickscatalog/bronze/autovol/destination/checkpoint")\
.option("cloudFiles.schemaEvolutionMode", "rescue")\
.load("/Volumes/databrickscatalog/bronze/autovol/raw/")
```

```python
df.writeStream.format("delta")\
    .option("checkpointLocation", "/Volumes/databrickscatalog/bronze/autovol/destination/checkpoint")\
    .outputMode("append")\
    .trigger(once=True)\
    .start("/Volumes/databrickscatalog/bronze/autovol/destination/data")
```

```python
##addnewColumn
df=spark.readStream.format("cloudFiles") \
.option("cloudFiles.format", "csv") \
.option("cloudFiles.schemaLocation","/Volumes/databrickscatalog/bronze/autovol/destination2/checkpoint")\
.option("cloudFiles.schemaEvolutionMode", "addNewColumns")\
.load("/Volumes/databrickscatalog/bronze/autovol/raw/")
```

```python
df.writeStream.format("delta")\
    .option("checkpointLocation", "/Volumes/databrickscatalog/bronze/autovol/destination2/checkpoint")\
    .outputMode("append")\
    .trigger(once=True)\
    .option("mergeSchema", "true")\
    .start("/Volumes/databrickscatalog/bronze/autovol/destination2/data")
```
