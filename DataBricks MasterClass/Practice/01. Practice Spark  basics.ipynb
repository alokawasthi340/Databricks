```python

data = [
    (1, 1, 2, "2023-01-10"),
    (2, 2, 2, "2023-01-11"),
    (3, 3, 1, "2023-01-12"),
    (4, 4, 5, "2023-01-13"),
    (5, 5, 5, "2023-01-14"),
    (6, 5, 5, "2023-01-15"),
    (7, 6, 3, "2023-01-16"),
    (8, 6, 6, "2023-01-17"),
    (9, 8, 7, "2023-01-18"),
    (10, 9, 10, "2023-01-19"),
    (11, 10, 10, "2023-01-20")
]

# Schema (column names)
columns = ["article_id", "author_id", "viewer_id", "view_date"]
df = spark.createDataFrame(data, columns)
display(df)
```

```python
result_df=(
    df
    .filter(df.author_id == df.viewer_id)
    .select("author_id")
    .distinct()
)
result_df.show()

# Learning distinct filter and select in pyspark
```

```python

# Data
data = [
    (1, "Will", None),
    (2, "Jane", None),
    (3, "Alex", 2),
    (4, "Bill", None),
    (5, "Zack", 1),
    (6, "Mark", 2)
]

# Column names
columns = ["id", "name", "referee_id"]

# Create DataFrame
customer_df = spark.createDataFrame(data, columns)
```

```python
customer_result_df=(
    customer_df.filter(
        customer_df.referee_id.isNull() | (customer_df.referee_id != 2)
    ).select("name")
)
customer_result_df.show()
```

```python

data = [
    (1, 500),
    (2, 800),
    (3, 800),
    (4, 600),
    (5, 300)
]

# Define column names
columns = ["item_count", "order_occurrences"]

# Create DataFrame
ItemsPerOrder_df = spark.createDataFrame(data, columns)

display(ItemsPerOrder_df)
```

```python
from pyspark.sql import functions as F
```

```python

max_val = ItemsPerOrder_df.agg(
    F.max("order_occurrences").alias("max_occ")
).collect()[0]["max_occ"]


ItemsPerOrderresult__df=(
    ItemsPerOrder_df
    .filter(ItemsPerOrder_df.order_occurrences ==max_val)
            .select("item_count").alias("mode")
)
ItemsPerOrderresult__df.show()
```
