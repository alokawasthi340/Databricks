```python
target_table=f"{catalog_name}.{gold_schema}.dim_race"
```

```python
circuit_df=spark.table(f"{catalog_name}.{silver_schema}.circuits")
race_df=spark.table(f"{catalog_name}.{silver_schema}.races")
```

```python
dim_race_final_df = race_df.join(
    circuit_df, race_df.circuit_id == circuit_df.circuit_id, "inner"
).select(
    race_df.race_name,
    race_df.race_date,
    race_df.season,
    race_df.round,
    circuit_df.circuit_name,
    circuit_df.country,
    circuit_df.locality,
)
```

```python

(
    dim_race_final_df.write.format("delta").mode("overwrite").saveAsTable(target_table)
)
```
