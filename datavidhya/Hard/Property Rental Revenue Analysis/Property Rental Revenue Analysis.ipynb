# Property Rental Revenue Analysis

## Problem Statement

A property manager wants the total monthly rent represented by each landlord's portfolio.

For every landlord with at least one property, return the landlord identifier, full name, and total rent across all properties.

## Input Tables

### prop_landlords

| Column Name | Data Type |
|------------|-----------|
| landlord_id | INT |
| first_name | VARCHAR |
| last_name | VARCHAR |
| email | VARCHAR |
| phone | VARCHAR |

### prop_properties

| Column Name | Data Type |
|------------|-----------|
| property_id | INT |
| landlord_id | INT |
| property_type | VARCHAR |
| rent | INT |
| square_feet | INT |
| city | VARCHAR |

## Requirements

- Include only landlords who own at least one property.
- Build `landlord_name` using first name, a single space, and last name.
- Return results ordered by `landlord_id` ascending.
- Return results matching the required output schema and order.

## Output Columns

| Column Name |
|------------|
| landlord_id |
| landlord_name |
| total_rental_income |

## Sample Input

### prop_landlords

| landlord_id | first_name | last_name | email | phone |
|------------|------------|-----------|--------|--------|
| 101 | John | Smith | john.smith@example.com | 555-123-4567 |
| 103 | Bob | Johnson | bob.johnson@example.com | 555-345-6789 |

### prop_properties

| property_id | landlord_id | property_type | rent | square_feet | city |
|------------|------------|---------------|------|-------------|------|
| 1 | 101 | Apartment | 1500 | 1000 | Seattle |
| 2 | 101 | Condo | 1200 | 800 | Seattle |
| 4 | 103 | Apartment | 1800 | 1200 | Redmond |
| 5 | 103 | Condo | 1000 | 700 | Redmond |

## Sample Output

| landlord_id | landlord_name | total_rental_income |
|------------|---------------|---------------------|
| 101 | John Smith | 2700 |
| 103 | Bob Johnson | 2800 |

## Expected Output Schema

| Column Name | Data Type |
|------------|-----------|
| landlord_id | INT |
| landlord_name | STRING |
| total_rental_income | INT |

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *

# prop_landlords
prop_landlords_schema = StructType([
    StructField("landlord_id", IntegerType(), True),
    StructField("first_name", StringType(), True),
    StructField("last_name", StringType(), True),
    StructField("email", StringType(), True),
    StructField("phone", StringType(), True)
])

prop_landlords_data = [
    (101, "John", "Smith", "john.smith@example.com", "555-123-4567"),
    (103, "Bob", "Johnson", "bob.johnson@example.com", "555-345-6789")
]

prop_landlords_df = spark.createDataFrame(
    prop_landlords_data,
    schema=prop_landlords_schema
)

# prop_properties
prop_properties_schema = StructType([
    StructField("property_id", IntegerType(), True),
    StructField("landlord_id", IntegerType(), True),
    StructField("property_type", StringType(), True),
    StructField("rent", IntegerType(), True),
    StructField("square_feet", IntegerType(), True),
    StructField("city", StringType(), True)
])

prop_properties_data = [
    (1, 101, "Apartment", 1500, 1000, "Seattle"),
    (2, 101, "Condo", 1200, 800, "Seattle"),
    (4, 103, "Apartment", 1800, 1200, "Redmond"),
    (5, 103, "Condo", 1000, 700, "Redmond")
]

prop_properties_df = spark.createDataFrame(
    prop_properties_data,
    schema=prop_properties_schema
)
```
