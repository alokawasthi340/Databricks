```sql
CREATE OR REPLACE TABLE prop_landlords (
  landlord_id INT,
  first_name STRING,
  last_name STRING,
  email STRING,
  phone STRING
);

INSERT INTO prop_landlords
  VALUES
    (101, 'John', 'Smith', 'john.smith@example.com', '555-123-4567'),
    (103, 'Bob', 'Johnson', 'bob.johnson@example.com', '555-345-6789');

CREATE OR REPLACE TABLE prop_properties (
  property_id INT,
  landlord_id INT,
  property_type STRING,
  rent INT,
  square_feet INT,
  city STRING
);

INSERT INTO prop_properties
  VALUES
    (1, 101, 'Apartment', 1500, 1000, 'Seattle'),
    (2, 101, 'Condo', 1200, 800, 'Seattle'),
    (4, 103, 'Apartment', 1800, 1200, 'Redmond'),
    (5, 103, 'Condo', 1000, 700, 'Redmond');

select
  pl.landlord_id,
  concat(pl.first_name, ' ', pl.last_name) as landlord_name,
  sum(pp.rent) as total_rent
from
  prop_landlords pl
    join prop_properties pp
      on pl.landlord_id = pp.landlord_id
group by
  pl.landlord_id,
  concat(pl.first_name, ' ', pl.last_name)
order by
  pl.landlord_id asc;
```
