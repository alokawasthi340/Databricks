```sql
CREATE OR REPLACE TABLE duplicate_product_listing (
  listing_id INT,
  vendor_id INT,
  product_name STRING,
  description STRING
);

INSERT INTO duplicate_product_listing
  VALUES
    (101, 501, 'Bluetooth Speaker', 'A portable speaker with Blueto'),
    (102, 501, 'Bluetooth Speaker', 'A portable speaker with Blueto'),
    (103, 502, 'Wireless Earbuds', 'Comfortable earbuds with high-'),
    (104, 502, 'Wireless Earbuds', 'Comfortable earbuds with high-');

select
  count(*) as duplicate_vendors
from
  (
    SELECT
      vendor_id,
      product_name,
      description
    FROM
      duplicate_product_listing
    GROUP BY
      vendor_id,
      product_name,
      description
    HAVING
      COUNT(*) > 1
  )
```
