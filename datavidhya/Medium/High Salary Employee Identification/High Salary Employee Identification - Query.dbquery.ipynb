```sql
CREATE OR REPLACE TABLE he_department (department_id INT, department_name STRING);

INSERT INTO he_department
  VALUES (1, 'Data Analytics'), (2, 'Data Science');

CREATE OR REPLACE TABLE he_employee (
  employee_id INT,
  name STRING,
  salary INT,
  department_id INT,
  manager_id DECIMAL(10, 0)
);

INSERT INTO he_employee
  VALUES
    (10, 'James Anderson', 4000, 1, 11),
    (1, 'Emma Thompson', 3800, 1, 6),
    (2, 'Daniel Rodriguez', 2230, 1, 7),
    (3, 'Olivia Smith', 2000, 1, 8),
    (4, 'Noah Johnson', 6800, 2, 9),
    (8, 'William Davis', 6800, 2, NULL);

with emp_with_rank as (
  select
    d.department_name,
    e.name,
    e.salary,
    dense_rank() over (partition by d.department_name order by e.salary desc, e.name asc) as rank
  from
    he_employee e
      join he_department d
        on e.department_id = d.department_id
)
select
  department_name,
  name,
  salary
from
  emp_with_rank
where
  rank <= 3
```
