```sql
CREATE OR REPLACE TABLE fr_customers (customer_id INT, full_name STRING, location STRING);

INSERT INTO fr_customers
  VALUES (1, 'John Doe', 'Texas'), (2, 'Jane Smith', 'California');

CREATE OR REPLACE TABLE fr_orders (order_id INT, customer_id INT, product_id INT, quantity INT);

INSERT INTO fr_orders
  VALUES (1001, 1, 101, 5), (1002, 2, 102, 2);

CREATE OR REPLACE TABLE fr_products (product_id INT, product_info STRING);

INSERT INTO fr_products
  VALUES (101, 'Carpet,Red'), (102, 'Tile,Blue');

-- SELECT explode(split('apple,banana,orange', ','));
select
  c.customer_id,
  split(c.full_name, ' ')[0] AS first_name,
  split(c.full_name, ' ')[1] AS last_name,
  c.location,
  o.order_id,
  split(p.product_info, ',')[0] AS product_type,
  split(p.product_info, ',')[1] AS product_color,
  p.product_id,
  o.quantity
from
  fr_customers c
    join fr_orders o
      on c.customer_id = o.customer_id
    join fr_products p
      on o.product_id = p.product_id;
```
