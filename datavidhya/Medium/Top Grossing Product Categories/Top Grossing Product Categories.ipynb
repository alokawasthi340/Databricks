# Top Grossing Product Categories

## Problem Statement

A retail team wants to identify the leading products in each category for the year 2022.

Analyze product spending data and return the top-performing products within each category based on the requirements provided.

## Input Table

### gi_product_spend

| Column Name | Data Type |
|------------|-----------|
| category | VARCHAR |
| product | VARCHAR |
| user_id | INT |
| spend | DECIMAL |
| transaction_date | TIMESTAMP |

## Requirements

- Consider transactions that occurred during 2022.
- Calculate spending totals at the category and product level.
- Determine product positions within each category based on total spending.
- Products with equal totals should share the same position.
- Return products occupying the top two positions within each category.
- Return results matching the required output schema and order.

## Output Columns

| Column Name |
|------------|
| category |
| product |
| total_spend |

## Sample Input

### gi_product_spend

| user_id | category | product | spend | transaction_date |
|----------|----------|---------|-------|------------------|
| 1 | appliance | fridge | 300 | 2022-01-01 |
| 2 | appliance | washer | 200 | 2022-02-01 |
| 3 | appliance | dryer | 200 | 2022-03-01 |
| 4 | electronics | phone | 250 | 2022-01-01 |
| 5 | electronics | tablet | 150 | 2022-02-01 |
| 6 | electronics | camera | 900 | 2021-12-31 |

## Sample Output

| category | product | total_spend |
|----------|---------|------------|
| appliance | fridge | 300 |
| appliance | dryer | 200 |
| appliance | washer | 200 |
| electronics | phone | 250 |
| electronics | tablet | 150 |

## Expected Output Schema

| Column Name | Data Type |
|------------|-----------|
| category | STRING |
| product | STRING |
| total_spend | DECIMAL |

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *

gi_product_spend_schema = StructType([
    StructField("user_id", IntegerType(), True),
    StructField("category", StringType(), True),
    StructField("product", StringType(), True),
    StructField("spend", DoubleType(), True),
    StructField("transaction_date", TimestampType(), True)
])

gi_product_spend_data = [
    (1, "appliance", "fridge", 300.0, "2022-01-01 00:00:00"),
    (2, "appliance", "washer", 200.0, "2022-02-01 00:00:00"),
    (3, "appliance", "dryer", 200.0, "2022-03-01 00:00:00"),
    (4, "electronics", "phone", 250.0, "2022-01-01 00:00:00"),
    (5, "electronics", "tablet", 150.0, "2022-02-01 00:00:00"),
    (6, "electronics", "camera", 900.0, "2021-12-31 00:00:00")
]

gi_product_spend_df = spark.createDataFrame(
    gi_product_spend_data,
    ["user_id", "category", "product", "spend", "transaction_date"]
).withColumn(
    "transaction_date",
    to_timestamp(col("transaction_date"))
)
```

```python
gi_product_spend_df=(
    gi_product_spend_df
    .filter(year(col("transaction_date"))==2022)
    .groupBy("category","product")
    .agg(
        sum("spend").alias("total_spend")
    )
)
display(gi_product_spend_df)
```
