```sql
CREATE OR REPLACE TABLE ttt_tweet_input (
    user_id INT,
    tweet_date TIMESTAMP,
    tweet_count INT
);

INSERT INTO ttt_tweet_input VALUES
(111, '2022-06-01 00:00:00', 2),
(111, '2022-06-02 00:00:00', 1),
(111, '2022-06-03 00:00:00', 3),
(111, '2022-06-04 00:00:00', 4),
(111, '2022-06-05 00:00:00', 5);


select 
user_id,
to_date(tweet_date,'yyyy-MM-dd') as tweet_date,
round(avg(tweet_count) over (partition by user_id order by tweet_date rows between 2 preceding and current row),2) as avg_tweet_count
from ttt_tweet_input
order by tweet_date,user_id 
```
