```sql
CREATE OR REPLACE TABLE ba_flights (
  flight_id INT,
  origin_airport STRING,
  destination_airport STRING,
  plane_id INT
);

INSERT INTO ba_flights
  VALUES (1, 'A1', 'B1', 2), (2, 'A2', 'B2', 3), (3, 'A3', 'B3', 1), (4, 'A9', 'B1', 2);

CREATE OR REPLACE TABLE ba_airports (airport_id STRING, airport_name STRING);

INSERT INTO ba_airports
  VALUES
    ('A1', 'San Francisco'),
    ('B1', 'Los Angeles'),
    ('A2', 'New York'),
    ('B2', 'Boston'),
    ('A3', 'Miami'),
    ('B3', 'Orlando');

CREATE OR REPLACE TABLE ba_planes (plane_id INT, plane_model STRING);

INSERT INTO ba_planes
  VALUES (1, 'Airbus A320'), (2, 'Boeing 737'), (3, 'Airbus A380');

select
  bf.flight_id,
  len(trim(ba1.airport_name)) as origin_airport_name_length,
  len(trim(ba2.airport_name)) as destination_airport_name_length,
  len(trim(bp.plane_model)) as plane_model_length
from
  ba_flights bf
    join ba_airports ba1
      on bf.origin_airport = ba1.airport_id
    join ba_airports ba2
      on bf.destination_airport = ba2.airport_id
    join ba_planes bp
      on bf.plane_id = bp.plane_id
```
