```sql
CREATE OR REPLACE TABLE pep_portfolio (
    pe_firm STRING,
    company STRING,
    shares INT
);

INSERT INTO pep_portfolio VALUES
('Alpha', 'A', 1000),
('Alpha', 'B', 2000),
('Beta', 'A', 1500),
('Beta', 'C', 2500),
('Gamma', 'B', 1200),
('Gamma', 'C', 1300);

CREATE OR REPLACE TABLE pep_prices (
    date DATE,
    company STRING,
    closing_price INT
);

INSERT INTO pep_prices VALUES
('2023-01-01', 'A', 50),
('2023-01-01', 'B', 20),
('2023-01-01', 'C', 30),
('2023-01-02', 'A', 52),
('2023-01-02', 'B', 21),
('2023-01-02', 'C', 31);

select 
  pe_firm as PE_firm, 
  to_char(date,'yyyy-MM-dd') as date,
  sum(shares*closing_price) as portfolio_value
from pep_portfolio pp
join pep_prices pr
on pp.company = pr.company
group by pe_firm, date
order by pe_firm;
```
