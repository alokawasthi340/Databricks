```sql
CREATE OR REPLACE TABLE mirmortgagedetails (
  MortgageID VARCHAR(10),
  MortgageType VARCHAR(20),
  InterestRate DECIMAL(10, 2)
);

INSERT INTO mirmortgagedetails (MortgageID, MortgageType, InterestRate)
  VALUES ('M1', 'Fixed', 4.50), ('M2', 'Variable', 3.20), ('M3', 'Adjustable', 2.80);

CREATE OR REPLACE TABLE mirusermortgages (UserID VARCHAR(10), MortgageID VARCHAR(10));

INSERT INTO mirusermortgages (UserID, MortgageID)
  VALUES ('U1', 'M1'), ('U2', 'M1'), ('U3', 'M2'), ('U4', 'M3');

with MortgageCte as (
  select
    MortgageType,
    sum(InterestRate) as TotalInterestRate,
    count(distinct UserID) as TotalUsers
  from
    mirmortgagedetails md
      join mirusermortgages mu
        on md.MortgageID = mu.MortgageID
  group by
    MortgageType
)
select
  MortgageType,
  round((TotalInterestRate / TotalUsers), 2) as AvgInterestRatePerUser
from
  MortgageCte;
```
