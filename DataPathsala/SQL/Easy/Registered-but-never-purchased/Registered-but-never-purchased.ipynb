```python
from pyspark.sql import SparkSession

# Initialize Spark
spark = SparkSession.builder.appName("UsersOrdersData").getOrCreate()

# Users DataFrame
users_data = [
    (1, "alice_w", "alice@dataco.com"),
    (2, "bob_t", "bob@dataco.com"),
    (3, "carol_m", "carol@dataco.com"),
    (4, "david_k", "david@dataco.com"),
    (5, "eve_r", "eve@dataco.com"),
]

users_df = spark.createDataFrame(users_data, ["id", "username", "email"])

# Orders DataFrame
orders_data = [
    (1, 1, "2023-03-10"),
    (2, 1, "2023-04-05"),
    (3, 3, "2023-05-20"),
    (4, 5, "2023-06-01"),
]

orders_df = spark.createDataFrame(orders_data, ["id", "user_id", "order_date"])

# Show DataFrames
print("Users DataFrame:")
users_df.show()

print("Orders DataFrame:")
orders_df.show()
```

```python
users_without_orders_df=users_df.join(orders_df, users_df.id==orders_df.user_id,"left_anti")
users_without_orders_df.show()

## left_anti means that we want to keep only the rows from the left DataFrame (users_df) that do not have a match in the right DataFrame (orders_df).
```
