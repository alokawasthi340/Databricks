```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f
from pyspark.sql.window import Window
# Initialize Spark
spark = SparkSession.builder.appName("OrdersData").getOrCreate()

# Create DataFrame
orders_data = [
    (1, 101, 120),
    (2, 101, 80),
    (3, 102, 45),
    (4, 102, 55),
    (5, 103, 300),
    (6, 104, 60),
    (7, 104, 40),
    (8, 105, 500),
    (9, 105, 200),
    (10, 106, 30),
]

orders_df = spark.createDataFrame(orders_data, ["id", "customer_id", "amount"])

# Show DataFrame
orders_df.show()
```

```python
result_df = (
    orders_df.groupBy("customer_id")
    .agg(f.sum("amount").alias("total_amount"))
    .withColumn("avg_total_amount", f.avg("total_amount").over(Window.partitionBy()))
    .select(
        "customer_id",
        "total_amount",
    )
    .filter(f.col("total_amount") > f.col("avg_total_amount"))
    .orderBy(f.col("total_amount").desc())
)
display(result_df)
```
