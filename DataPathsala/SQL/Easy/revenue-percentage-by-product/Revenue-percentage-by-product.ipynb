```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f  
from pyspark.sql.window import Window
# Initialize Spark
spark = SparkSession.builder.appName("SalesData").getOrCreate()

# Create DataFrame
sales_data = [
    (1, "Laptop Pro", 1200.0),
    (2, "Wireless Mouse", 45.0),
    (3, "Laptop Pro", 950.0),
    (4, "USB-C Hub", 89.99),
    (5, "Mechanical Keyboard", 149.0),
    (6, "Wireless Mouse", 45.0),
    (7, "Laptop Pro", 1100.0),
    (8, "USB-C Hub", 89.99),
    (9, "Mechanical Keyboard", 149.0),
    (10, "Mechanical Keyboard", 149.0),
    (11, "Wireless Mouse", 60.0),
]

sales_df = spark.createDataFrame(sales_data, ["id", "product_name", "revenue"])

# Show DataFrame
sales_df.show()
```

```python
result_df = (
    sales_df.groupBy("product_name")
    .agg(f.sum("revenue").alias("total_revenue"))
    .withColumn("total_grand", f.sum("total_revenue").over(Window.partitionBy()))
    .withColumn(
        "revenue_pct", f.round(f.col("total_revenue") * 100 / f.col("total_grand"), 2)
    )
    .select(f.col("product_name"), f.col("total_revenue"), f.col("revenue_pct"))
    .orderBy(f.col("total_revenue").desc())
)
display(result_df)
```
