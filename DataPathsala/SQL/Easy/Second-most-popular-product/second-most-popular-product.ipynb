```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f 

# Initialize Spark
spark = SparkSession.builder.appName("OrderItemsData").getOrCreate()

# Create DataFrame
order_items_data = [
    (1, 1, 101, 10),
    (2, 1, 102, 5),
    (3, 2, 101, 3),
    (4, 2, 103, 20),
    (5, 3, 102, 8),
    (6, 3, 104, 13),
    (7, 4, 103, 7),
    (8, 4, 105, 2),
]

order_items_df = spark.createDataFrame(order_items_data, ["id", "order_id", "product_id", "quantity"])

# Show DataFrame
order_items_df.show()
```

```python
from pyspark.sql import Window

window = Window.orderBy(f.sum(f.col("quantity")).desc())

second_product_df = (
    order_items_df.groupBy("product_id")
    .agg(
        f.sum("quantity").alias("total_quantity"),
        f.dense_rank().over(window).alias("rank"),
    )
    .select("product_id", "total_quantity")
    .filter(f.col("rank") == 2)
)
display(second_product_df)
```
