```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f
# Create Spark session
spark = SparkSession.builder.appName("CoursesDataFrame").getOrCreate()

# Define data
data = [
    ("A", "Math"),
    ("B", "English"),
    ("C", "Math"),
    ("D", "Biology"),
    ("E", "Math"),
    ("F", "Math"),
    ("G", "Math"),
    ("H", "English"),
    ("I", "English"),
    ("J", "English"),
    ("K", "English")
]

# Define schema (column names)
columns = ["student", "class"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Show DataFrame
df.show()
```

```python
df_result=(
    df
    .groupBy("class")
    .agg(
        f.count("*").alias("total_students")
    )
    .filter("total_students > 4")
    .select("class")
)
display(df_result)
```
