```python
from pyspark.sql import SparkSession

# Initialize Spark
spark = SparkSession.builder.appName("EmployeesData").getOrCreate()

# Create DataFrame
employees_data = [
    (1, "Sarah Connor", "Engineering", None),
    (2, "James Reyes", "Engineering", 1),
    (3, "Priya Patel", "Marketing", 1),
    (4, "Tom Bradley", "Sales", 3),
    (5, "Nina Watts", "Marketing", 3),
    (6, "Omar Hassan", "Engineering", 2),
    (7, "Lily Zhang", "Sales", 4),
]

employees_df = spark.createDataFrame(employees_data, ["id", "name", "department", "manager_id"])

# Show DataFrame
employees_df.show()
```

```python
emp_df = employees_df.alias("emp")
mgr_df = employees_df.alias("mgr")

result_df=(
    emp_df.join(mgr_df, emp_df.manager_id == mgr_df.id, "inner")
    .filter(emp_df.department != mgr_df.department)
    .select(
        emp_df.name.alias("employee_name"), emp_df.department.alias("employee_department"), mgr_df.name.alias("manager_name"), mgr_df.department.alias("manager_department"))
)

display(result_df)
```
