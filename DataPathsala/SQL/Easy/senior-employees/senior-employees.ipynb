```python
from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder.appName("EmployeesDataFrame").getOrCreate()

# Input data
data = [
    (1, "Alice Smith", "Engineering", 12, 130000),
    (2, "Bob Johnson", "Marketing", 5, 75000),
    (3, "Carol White", "Engineering", 15, 155000),
    (4, "David Brown", "Sales", 8, 85000),
    (5, "Eva Martinez", "HR", 10, 95000)
]

# Column names
columns = ["emp_id", "name", "department", "years_experience", "salary"]

# Create DataFrame
df_employees = spark.createDataFrame(data, columns)

# Show DataFrame
df_employees.show()
```

```python
result_df = (
    df_employees.filter(df_employees.years_experience >= 10)
    .orderBy("years_experience", ascending=False)
    .select("emp_id", "name", "years_experience")
)
display(result_df)
```
