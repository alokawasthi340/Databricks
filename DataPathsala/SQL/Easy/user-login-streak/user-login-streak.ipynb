```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f 
# Initialize Spark
spark = SparkSession.builder.appName("LoginsData").getOrCreate()

# Create DataFrame
logins_data = [
    (1, "alice_w", "2024-01-01"),
    (2, "alice_w", "2024-01-02"),
    (3, "alice_w", "2024-01-05"),
    (4, "bob_k", "2024-01-10"),
    (5, "bob_k", "2024-01-11"),
    (6, "bob_k", "2024-01-12"),
    (7, "carol_m", "2024-01-15"),
    (8, "carol_m", "2024-01-17"),
]

logins_df = spark.createDataFrame(logins_data, ["id", "user_name", "login_date"])
logins_df = logins_df.withColumn("login_date", f.to_date(f.col("login_date")))
# Show DataFrame
logins_df.show()
```

```python
l1_df = logins_df.alias("l1")
l2_df = logins_df.alias("l2")

result_df = (
    l1_df.join(l2_df, f.datediff(l1_df.login_date, l2_df.login_date) == 1, "inner")
    .select(l1_df.user_name.alias("user_name"))
    .distinct()
)
display(result_df)
```
