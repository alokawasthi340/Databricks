```python
from pyspark.sql import SparkSession

# Initialize Spark
spark = SparkSession.builder.appName("SalesData").getOrCreate()

# Create DataFrame
sales_data = [
    (1, "2023-06-01", 250),
    (2, "2023-06-01", 150),
    (3, "2023-06-02", 300),
    (4, "2023-06-03", 100),
    (5, "2023-06-03", 200),
    (6, "2023-06-04", 500),
]

sales_df = spark.createDataFrame(sales_data, ["id", "sale_date", "amount"])

# Show DataFrame
sales_df.show()
```

```python
from pyspark.sql import functions as f
from pyspark.sql import Window

windowSpec = Window.orderBy("sale_date")
result_df = (
    sales_df.groupBy("sale_date").agg(
        f.sum("amount").alias("total_sales"),
    )
).withColumn("running_total", f.sum("total_sales").over(windowSpec))
display(result_df)
```
