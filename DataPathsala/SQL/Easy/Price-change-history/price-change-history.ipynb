```python
from pyspark.sql import SparkSession

# Initialize Spark
spark = SparkSession.builder.appName("PriceHistory").getOrCreate()

# Create DataFrame
price_data = [
    (1, 1001, 19.99, "2022-01-01"),
    (2, 1001, 24.99, "2022-07-15"),
    (3, 1001, 21.99, "2023-01-10"),
    (4, 1002, 5.49, "2022-03-01"),
    (5, 1002, 6.99, "2023-06-01"),
    (6, 1003, 149.0, "2021-11-01"),
    (7, 1003, 129.0, "2022-11-01"),
    (8, 1003, 139.0, "2023-11-01"),
    (9, 1004, 9.99, "2023-08-01"),
]

price_df = spark.createDataFrame(price_data, ["id", "product_id", "price", "effective_date"])

# Show DataFrame
price_df.show()
```

```python
from pyspark.sql.window import Window
import pyspark.sql.functions as f

windowSpec = Window.partitionBy("product_id").orderBy(f.col("effective_date").desc())

result_df = (
    price_df.withColumn("rank", f.dense_rank().over(windowSpec))
    .filter(f.col("rank") == 1)
    .select("product_id", "price", "effective_date")
)
display(result_df)
```
