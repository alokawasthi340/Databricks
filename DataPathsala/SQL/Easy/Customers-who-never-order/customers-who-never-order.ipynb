```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f

# Initialize Spark
spark = SparkSession.builder.appName("CustomerOrders").getOrCreate()

# Create customers DataFrame
customers_data = [(1, "Joe"), (2, "Henry"), (3, "Sam"), (4, "Max")]
customers_df = spark.createDataFrame(customers_data, ["id", "name"])

# Create orders DataFrame
orders_data = [(1, 3), (2, 1)]
orders_df = spark.createDataFrame(orders_data, ["id", "customer_id"])

display(customers_df)
display(orders_df)
```

```python
result_df = (
    customers_df.join(orders_df, customers_df.id == orders_df.customer_id, "left")
    .filter(orders_df.id.isNull())
    .select("name")
)
display(result_df)
```
