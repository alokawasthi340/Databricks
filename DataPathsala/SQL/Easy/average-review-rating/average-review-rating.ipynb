```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import to_date
from pyspark.sql import functions as f
# Initialize Spark session
spark = SparkSession.builder.appName("ReviewsDataFrame").getOrCreate()

# Input data
data = [
    (1, 101, 1, 5, "2023-01-10"),
    (2, 101, 2, 4, "2023-01-11"),
    (3, 102, 3, 3, "2023-01-12"),
    (4, 102, 4, 3, "2023-01-13"),
    (5, 103, 5, 5, "2023-01-14"),
    (6, 103, 6, 5, "2023-01-15"),
    (7, 103, 7, 4, "2023-01-16")
]

# Column names
columns = ["review_id", "product_id", "user_id", "rating", "review_date"]

# Create DataFrame
df_reviews = spark.createDataFrame(data, columns)

# Convert review_date to DateType
df_reviews = df_reviews.withColumn("review_date", to_date("review_date", "yyyy-MM-dd"))

# Show DataFrame
df_reviews.show()
```

```python
result_df = (
    df_reviews.groupBy("product_id")
    .agg(f.avg("rating").alias("avg_rating"))
    .orderBy(f.col("avg_rating").desc(), f.col("product_id").asc())
)
display(result_df)
```
