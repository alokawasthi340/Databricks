```python
from pyspark.sql import SparkSession

# Initialize Spark
spark = SparkSession.builder.appName("DocumentsData").getOrCreate()

# Create DataFrame
documents_data = [
    (1, "the cat sat on the mat"),
    (2, "the dog sat on the log"),
    (3, "the cat and the dog"),
]

documents_df = spark.createDataFrame(documents_data, ["doc_id", "text"])

# Show DataFrame
documents_df.show()
```

```python
from pyspark.sql import functions as F

result_df = (
    documents_df.withColumn("word", F.explode(F.split(F.lower(F.col("text")), " ")))
    .groupBy(F.col("word"))
    .agg(F.count("*").alias("count"))
    .orderBy(F.col("count").desc(), F.col("word").asc())
)
display(result_df)
```
