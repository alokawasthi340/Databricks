```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f
# Create Spark session
spark = SparkSession.builder.appName("WeatherDataFrame").getOrCreate()

# Define data
data = [
    (1, "2015-01-01", 10),
    (2, "2015-01-02", 25),
    (3, "2015-01-03", 20),
    (4, "2015-01-04", 30)
]

# Define column names
columns = ["id", "record_date", "temperature"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Show DataFrame
df.show()
```

```python
df_left = df.alias("w")
df_right = df.alias("y")
result_df = (
    df_left.join(
        df_right,
        f.datediff(f.col("w.record_date"), f.col("y.record_date")) == 1,
    )
    .filter(f.col("w.temperature") > f.col("y.temperature"))
    .select(f.col("w.id").alias("id"))
)
display(result_df)
```
