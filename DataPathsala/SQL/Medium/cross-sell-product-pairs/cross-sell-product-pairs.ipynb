```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f
# Initialize Spark session
spark = SparkSession.builder.appName("OrderItemsDF").getOrCreate()

# Define the dataset
data = [
    (1, 1001, 1, "Laptop"),
    (2, 1001, 2, "Mouse"),
    (3, 1001, 3, "Keyboard"),
    (4, 1002, 1, "Laptop"),
    (5, 1002, 2, "Mouse"),
    (6, 1003, 1, "Laptop"),
    (7, 1003, 2, "Mouse"),
    (8, 1003, 3, "Keyboard"),
    (9, 1004, 2, "Mouse"),
    (10, 1004, 3, "Keyboard"),
    (11, 1005, 1, "Laptop"),
    (12, 1005, 3, "Keyboard"),
    (13, 1006, 1, "Laptop"),
    (14, 1006, 2, "Mouse"),
    (15, 1006, 3, "Keyboard"),
    (16, 1007, 2, "Mouse"),
    (17, 1007, 3, "Keyboard")
]

# Define schema
columns = ["id", "order_id", "product_id", "product_name"]

# Create DataFrame
order_items_df = spark.createDataFrame(data, columns)

# Show DataFrame
order_items_df.show()
```

```python
ot_df = order_items_df.alias("ot")
os_df = order_items_df.alias("os")

result_df = (
    ot_df.join(
        os_df,
        (f.col("ot.order_id") == f.col("os.order_id"))
        & (f.col("ot.product_id") < f.col("os.product_id")),
    )
    .groupBy(f.col("ot.product_name"), f.col("os.product_name"))
    .agg(f.count("*").alias("frequency"))
    .select(
        f.col("ot.product_name").alias("product_name_1"),
        f.col("os.product_name").alias("product_name_2"),
        f.col("frequency"),
    )
    .orderBy(f.col("frequency").desc())
)
display(result_df)
```
