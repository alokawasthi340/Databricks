```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f

# Initialize Spark
spark = SparkSession.builder.appName("ViewsDataset").getOrCreate()

# === Views DataFrame ===
views_data = [
    (1, 3, 5, "2019-08-01"),
    (3, 4, 5, "2019-08-01"),
    (1, 3, 6, "2019-08-02"),
    (2, 7, 7, "2019-08-01"),
    (2, 7, 6, "2019-08-02"),
    (4, 7, 1, "2019-07-22"),
    (3, 4, 4, "2019-07-21"),
    (3, 4, 4, "2019-07-21")
]

views_df = spark.createDataFrame(
    views_data,
    ["article_id", "author_id", "viewer_id", "view_date"]
)

# Show the DataFrame
views_df.show()
```

```python
result_df = (
    views_df.groupBy("viewer_id", "view_date")
    .agg(f.countDistinct("article_id").alias("articles_viewed"))
    .filter(f.col("articles_viewed") > 1)
    .select("viewer_id")
    .distinct()
)
display(result_df)
```
