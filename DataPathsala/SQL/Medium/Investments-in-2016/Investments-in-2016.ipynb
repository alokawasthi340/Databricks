```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f
# Initialize Spark session
spark = SparkSession.builder.appName("InsuranceData").getOrCreate()

# Define the dataset
data = [
    (1, 10, 5, 10, 10),
    (2, 20, 20, 20, 20),
    (3, 10, 30, 20, 20),
    (4, 10, 40, 40, 40)
]

# Define the schema
columns = ["pid", "tiv_2015", "tiv_2016", "lat", "lon"]

# Create DataFrame
insurance_df = spark.createDataFrame(data, columns)

# Show DataFrame
insurance_df.show()
```

```python
result_df = (
    insurance_df.select((f.sum(insurance_df.tiv_2016).alias("tiv_2016")))
).where(
    f.col("tiv_2015")
    == (
        insurance_df.groupBy("tiv_2015")
        .agg(f.sum(insurance_df.tiv_2015).alias("tiv_2015_count"))
        .filter(f.col("tiv_2015_count") > 1)
        .select(f.col("tiv_2015"))
    )
)
```
