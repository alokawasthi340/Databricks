```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f
from pyspark.sql import Window

# Initialize Spark session
spark = SparkSession.builder.appName("OrdersData").getOrCreate()

# Define the dataset
data = [
    (1, "2024-02-01 08:15:00", 320),
    (2, "2024-02-01 08:45:00", 180),
    (3, "2024-02-01 12:05:00", 540),
    (4, "2024-02-01 12:30:00", 610),
    (5, "2024-02-01 12:55:00", 290),
    (6, "2024-02-01 18:10:00", 750),
    (7, "2024-02-01 18:40:00", 820),
    (8, "2024-02-02 08:20:00", 400),
    (9, "2024-02-02 12:10:00", 350),
    (10, "2024-02-02 18:50:00", 270),
    (11, "2024-02-02 21:00:00", 900),
    (12, "2024-02-02 21:30:00", 600)
]

# Define schema
columns = ["id", "order_time", "amount"]

# Create DataFrame
orders_df = spark.createDataFrame(data, columns)

# Convert order_time to TimestampType
orders_df = orders_df.withColumn("order_time", f.col("order_time").cast("timestamp"))

# Show DataFrame
orders_df.show(truncate=False)
```

```python
result_df = (
    orders_df.withColumn("hour_of_day", f.hour(f.col("order_time")))
    .groupBy("hour_of_day")
    .agg(f.sum("amount").alias("total_amount"), f.count("*").alias("order_count"))
    .orderBy(f.desc("total_amount"))
    .limit(3)
)
display(result_df)
```
