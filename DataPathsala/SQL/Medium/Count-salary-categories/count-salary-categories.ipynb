```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f

# Initialize Spark session
spark = SparkSession.builder.appName("AccountsData").getOrCreate()

# Define dataset
accounts_data = [
    (1, 15000),
    (2, 25000),
    (3, 60000),
    (4, 45000),
    (5, 8000)
]

# Define schema
accounts_columns = ["account_id", "income"]

# Create DataFrame
accounts_df = spark.createDataFrame(accounts_data, accounts_columns)

# Show DataFrame
accounts_df.show()
```
