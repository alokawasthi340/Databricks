```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DateType
from pyspark.sql import functions as f
# Initialize Spark session
spark = SparkSession.builder.appName("EmailsTextsExample").getOrCreate()

# Define schema for emails
emails_schema = StructType([
    StructField("email_id", IntegerType(), True),
    StructField("user_id", IntegerType(), True),
    StructField("signup_date", StringType(), True)  # can later cast to DateType
])

# Create emails DataFrame
emails_data = [
    (1, 1001, "2023-01-01"),
    (2, 1002, "2023-01-02"),
    (3, 1003, "2023-01-03"),
    (4, 1004, "2023-01-04"),
    (5, 1005, "2023-01-05")
]

emails_df = spark.createDataFrame(emails_data, schema=emails_schema)

# Define schema for texts
texts_schema = StructType([
    StructField("text_id", IntegerType(), True),
    StructField("email_id", IntegerType(), True),
    StructField("signup_action", StringType(), True)
])

# Create texts DataFrame
texts_data = [
    (10, 1, "Confirmed"),
    (11, 2, "Not Confirmed"),
    (12, 3, "Confirmed"),
    (13, 5, "Confirmed")
]

texts_df = spark.createDataFrame(texts_data, schema=texts_schema)

# Show the DataFrames
print("Emails DataFrame:")
emails_df.show()

print("Texts DataFrame:")
texts_df.show()
```

```python
result_df = emails_df.join(texts_df, on="email_id", how="left").select(
    f.round(
        f.sum(f.when(f.col("signup_action") == "Confirmed", 1).otherwise(f.lit(0)))
        / f.count("*"),
        2,
    ).alias("activation_rate")
)
display(result_df)
```
