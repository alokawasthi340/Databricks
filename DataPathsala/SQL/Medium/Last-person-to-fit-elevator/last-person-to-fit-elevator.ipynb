```python
%sql
-- Create the table
CREATE TABLE queue (
    person_id INT PRIMARY KEY,
    person_name VARCHAR(50),
    weight INT,
    turn INT
);

-- Insert the data
INSERT INTO queue (person_id, person_name, weight, turn) VALUES
(1, 'Oscar', 100, 1),
(2, 'Pat', 150, 2),
(3, 'Quinn', 200, 3),
(4, 'Rita', 250, 4),
(5, 'Sam', 300, 5),
(6, 'Tina', 50, 6);
```

```python
%sql
SELECT
  person_id,
  person_name,
  weight,
  turn,
  sum(weight) over (order by turn) as running_sum
FROM
  queue
```

```python
%sql
SELECT
  person_id,
  person_name,
  weight,
  turn,
  sum(weight) over (order by turn) as running_sum
FROM
  queue
```

```python
%sql
-- Write your SQL query here
SELECT
  person_name
from
  (
    SELECT
      person_id,
      person_name,
      weight,
      turn,
      sum(weight) over (order by turn) as running_sum
    FROM
      queue
  ) x
where
  x.running_sum <= 1000
  order by x.running_sum desc
limit 1
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f
from pyspark.sql.window import Window
# Initialize Spark session
spark = SparkSession.builder.appName("QueueData").getOrCreate()

# Define the dataset
data = [
    (1, "Oscar", 100, 1),
    (2, "Pat", 150, 2),
    (3, "Quinn", 200, 3),
    (4, "Rita", 250, 4),
    (5, "Sam", 300, 5),
    (6, "Tina", 50, 6)
]

# Define schema
columns = ["person_id", "person_name", "weight", "turn"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Show the DataFrame
df.show()
```

```python
result_df = (
    df.withColumn("running_sum", f.sum("weight").over(Window.orderBy("turn")))
    .filter(f.col("running_sum") <= 1000)
    .select("person_name")
    .orderBy(f.desc(f.col("running_sum")))
    .limit(1)
)
display(result_df)
```
