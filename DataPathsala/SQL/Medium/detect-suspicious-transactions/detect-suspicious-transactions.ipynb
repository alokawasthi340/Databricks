```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f
from pyspark.sql import Window
# Initialize Spark session
spark = SparkSession.builder.appName("TransactionsDF").getOrCreate()

# Transactions dataset
transactions_data = [
    (301, 404, 135, "2024-01-03"),
    (302, 404, 165, "2024-01-10"),
    (303, 404, 150, "2024-01-17"),
    (304, 404, 1950, "2024-01-24"),
    (305, 408, 600, "2024-01-05")
]

transactions_columns = ["id", "user_id", "amount", "transaction_date"]

# Create DataFrame
transactions_df = spark.createDataFrame(transactions_data, transactions_columns)

# Show DataFrame
transactions_df.show()
```

```python
result_df = (
    transactions_df.withColumn(
        "user_avg_amount", f.avg("amount").over(Window.partitionBy("user_id"))
    )
    .filter(3 * f.col("user_avg_amount") <= transactions_df.amount)
    .select(
        f.col("id").alias("transaction_id"),
        f.col("user_id"),
        f.col("amount"),
        f.col("user_avg_amount"),
    )
    .distinct()
)
display(result_df)
```
