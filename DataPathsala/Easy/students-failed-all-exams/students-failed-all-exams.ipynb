```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, StringType
from pyspark.sql import functions
# Initialize Spark
spark = SparkSession.builder.appName("StudentsExamsDF").getOrCreate()

# -------------------------
# Students DataFrame
# -------------------------
students_schema = StructType([
    StructField("student_id", IntegerType(), True),
    StructField("student_name", StringType(), True)
])

students_data = [
    (1, "Alice"),
    (2, "Bob"),
    (3, "Charlie"),
    (4, "Diana")
]

students_df = spark.createDataFrame(students_data, schema=students_schema)

# -------------------------
# Exams DataFrame
# -------------------------
exams_schema = StructType([
    StructField("exam_id", IntegerType(), True),
    StructField("student_id", IntegerType(), True),
    StructField("score", IntegerType(), True)
])

exams_data = [
    (1, 1, 45),
    (2, 1, 30),
    (3, 1, 55),
    (4, 2, 70),
    (5, 2, 40),
    (6, 3, 20),
    (7, 3, 15),
    (8, 4, 80),
    (9, 4, 90)
]

exams_df = spark.createDataFrame(exams_data, schema=exams_schema)

# Show DataFrames
students_df.show()
exams_df.show()
```

```python
result_df = (
    students_df.join(exams_df, students_df.student_id == exams_df.student_id)
    .groupBy(students_df.student_id, students_df.student_name)
    .agg(
        functions.sum(functions.when(functions.col("score") >= 60, 1)).alias("qualified")
    )
    .filter(functions.col("qualified").isNull())
    .select(students_df.student_id,students_df.student_name) 
)
display(result_df)
```
