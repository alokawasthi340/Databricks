```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, FloatType
from pyspark.sql import functions as F
# Initialize Spark session
spark = SparkSession.builder.appName("PharmacySalesDataFrame").getOrCreate()

# Input data (use float values explicitly)
data = [
    (1, "Humira", 100000.0, 50000.0),
    (2, "Keytruda", 80000.0, 30000.0),
    (3, "Dupixent", 60000.0, 25000.0),
    (4, "Eliquis", 45000.0, 40000.0),
    (5, "Stelara", 70000.0, 55000.0)
]

# Define schema with FloatType
schema = StructType([
    StructField("product_id", IntegerType(), True),
    StructField("drug", StringType(), True),
    StructField("total_sales", FloatType(), True),
    StructField("cogs", FloatType(), True)
])

# Create DataFrame
df_pharmacy = spark.createDataFrame(data, schema)

# Show DataFrame
df_pharmacy.show()
```

```python
result_df = (
    df_pharmacy.select(
        "drug", (F.col("total_sales") - F.col("cogs")).alias("total_profit")
    )
    .orderBy(F.col("total_profit").desc())
    .limit(3)
)
display(result_df)
```
