```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import to_timestamp

# Initialize Spark session
spark = SparkSession.builder.appName("PostsDataFrame").getOrCreate()

# Input data
data = [
    (1, 101, "2024-01-05 10:00:00"),
    (1, 102, "2024-03-20 14:30:00"),
    (1, 103, "2024-06-15 09:00:00"),
    (2, 104, "2024-02-10 11:00:00"),
    (3, 105, "2024-04-01 08:00:00"),
    (3, 106, "2024-04-10 16:00:00"),
    (4, 107, "2023-12-31 23:59:59")
]

# Column names
columns = ["user_id", "post_id", "post_date"]

# Create DataFrame
df_posts = spark.createDataFrame(data, columns)

# Convert to TimestampType
df_posts = df_posts.withColumn(
    "post_date", 
    to_timestamp("post_date", "yyyy-MM-dd HH:mm:ss")
)

# Show DataFrame
df_posts.show(truncate=False)
```
