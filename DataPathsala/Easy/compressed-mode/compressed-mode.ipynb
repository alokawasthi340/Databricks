```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f
# Initialize Spark session
spark = SparkSession.builder.appName("ItemsPerOrderDataFrame").getOrCreate()

# Input data
data = [
    (1, 500),
    (2, 800),
    (3, 800),
    (4, 600),
    (5, 300)
]

# Column names
columns = ["item_count", "order_occurrences"]

# Create DataFrame
df_items = spark.createDataFrame(data, columns)

# Show DataFrame
df_items.show()
```

```python
max_df = df_items.agg(f.max("order_occurrences").alias("max_occ"))
result_df = (
    df_items
    .join(max_df, df_items.order_occurrences == max_df.max_occ)).select(
    "item_count"
)
display(result_df)
```
