```python
from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder.appName("ActivityDataFrame").getOrCreate()

# Sample data
data = [
    (1, 2, "2023-01-01", 5),
    (1, 2, "2023-01-05", 6),
    (2, 3, "2023-01-02", 1),
    (3, 1, "2023-01-03", 0),
    (3, 4, "2023-01-01", 2)
]

# Define schema (column names)
columns = ["player_id", "device_id", "event_date", "games_played"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Optional: cast event_date to DateType
from pyspark.sql.functions import col
df = df.withColumn("event_date", col("event_date").cast("date"))

# Show DataFrame
df.show()
```

```python
from pyspark.sql import Window
import pyspark.sql.functions as f

# 1. Define the window
windowSpec = Window.partitionBy("player_id").orderBy("event_date")

result_df=(
    df.withColumn("rank",f.dense_rank().over(windowSpec))
    .filter(f.col("rank")==1)
    .select("player_id",f.col("event_date").alias("first_login"))
    .orderBy("player_id")
)
result_df.show()
```
